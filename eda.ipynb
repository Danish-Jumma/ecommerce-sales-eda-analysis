### Clean Price

```python
df['price'] = (df['price'].str.replace('$' , '', regex=False).str.replace(',', '', regex=False)).astype(float)
```

Clean Discount Column

```python
df['discount']
```

```python
df['discount'] = df['discount'].str.replace('%', '', regex=False)
df['discount'] = df['discount'].str.replace('-', '', regex=False)
df['discount'] = pd.to_numeric(df['discount'], errors= 'coerce')
```

```python
df['discount']
```

Handle Missing Values

```python
df.isnull().sum()
```

```python
df['discount'] = df['discount'].fillna(0)
```

```python
df['discount']
```

```python
df['rank-title']
```

```python
df['rank-title'] = df['rank-title'].fillna('No Rank')
```

```python
df['rank-title']
```

Exract Sales Number

```python
df['selling_proposition']
```

```python
df['sold_count'] = df['selling_proposition'].str.extract(r'([\d\.]+k?)')
```

```python
df['sold_count']
```

```python
def convert_k(x):
    if pd.isna(x):
        return 0   

    x = str(x)    

    if 'k' in x:
        return float(x.replace('k', '')) * 1000

    return float(x)

df['sold_count'] = df['sold_count'].apply(convert_k)
```

```python
df['sold_count']
```

```python
df.columns
```

```python
df.drop(columns=['selling_proposition'], inplace=True)
```

```python
df.columns
```

Clean Product Title

```python
df['product_title'] = df['product_title'].fillna('Unknown Product')
```
